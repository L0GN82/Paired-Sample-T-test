```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import adjusted_rand_score
from itertools import combinations
```

```python
main_df = pd.read_excel(r"C:\Users\LoganFrideley\East Texas Salt Water Disposal Company\Operations - Documents\Engineering\Disposal Wells\Workover Analysis\Derived All Data.xlsx")
print(main_df.columns)
```

```python
candidates_df = main_df[["Billing KWH", "Electrical Bill", "Metered DM", "PSI", "Water Total", "Electrical Cost per Barrel"]]
candidates_df
```

```python
cols = ["Billing KWH",
        "Electrical Bill",
        "Metered DM",
        "PSI",
        "Water Total",
        "Electrical Cost per Barrel"]

for comb in combinations(cols, 2):
    exog1, exog2 = comb
    plt.figure(figsize=(7, 5))
    plt.scatter(candidates_df[exog1], candidates_df[exog2])

    plt.xlabel(exog1)
    plt.ylabel(exog2)
    plt.title(f"{exog1} vs {exog2}")

    plt.show()
    
        
        
```

```python
cols = ["Billing KWH",
        "Electrical Bill",
        "Metered DM",
        "PSI",
        "Water Total",
        "Electrical Cost per Barrel"]

for comb in combinations(cols, 3):
    exog1, exog2, exog3 = comb
    

    fig = plt.figure(figsize=(9, 7))
    ax = fig.add_subplot(111, projection='3d')
    
    scatter = ax.scatter(
        candidates_df[exog1],
        candidates_df[exog2],
        candidates_df[exog3],
        cmap='viridis'
    )
    
    ax.set_xlabel(exog1)
    ax.set_ylabel(exog2)
    ax.set_zlabel(exog3)

    fig.colorbar(scatter, ax=ax, pad=0.1)
    plt.title('3D Scatter Plot')
    plt.show()
```
