```python
candidates_df = main_df[["Billing KWH", "Electrical Bill", "Metered DM", "PSI", "Power Factor", "Water Total", "Electrical Cost per Barrel"]]
candidates_df
```

```python
#Replacing the Nan pump model values with valid entries
filt_df = main_df[["Name", "Pump Model"]]
na_filt = filt_df["Pump Model"].isna()
wells_and_pumps = filt_df[~na_filt]

# 1. Create a lookup mapping from Name -> Pump Model
pump_lookup = wells_and_pumps.drop_duplicates("Name").set_index("Name")["Pump Model"]

# 2. Map the lookup values across the entire 'Name' column
mapped_pumps = main_df["Name"].map(pump_lookup)

# 3. Use np.where to fill NaNs
main_df["Pump Model"] = np.where(
    main_df["Pump Model"].isna(), 
    mapped_pumps, 
    main_df["Pump Model"]
)

main_df["Pump Model"]
```
